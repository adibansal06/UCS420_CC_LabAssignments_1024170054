Computers cannot generate truly random numbers using software alone. Instead, Python's random module uses an algorithm called the Mersenne Twister to generate a sequence of numbers that appear random.

The Seed is the Starting Value: The integer 18932749 serves as the initial mathematical key that sets the state of the generator.

Deterministic Output: Every time you run your script with random.seed(18932749), Python will generate the exact same sequence of numbers in the exact same order.

```python
import random as r
r.seed(1024170054)
random_numbers = [r.randint(100,900) for _ in range(100)]
# here in randint() upper bound is also included, also _ here mean that the we are just iterating, nothing special, same as some other variable
print(random_numbers)
odd_random_numbers = [i for i in random_numbers if (i%2 == 1)]
print(odd_random_numbers)
count_odd = len(odd_random_numbers)
print(count_odd)
even_random_numbers = [i for i in random_numbers if (i%2 == 0)]
print(even_random_numbers)
count_even = sum(1 for i in random_numbers if i%2 == 0)
count_even1 = sum([1 for i in random_numbers if i%2 == 0])
print(count_even)
print(count_even1)

def isPrime(x) :
  count = 0
  for i in range(1,x+1):
    if x%i == 0:
      count += 1
  if count == 2:
    return True
  else:
    return False


prime_random_numbers = [i for i in random_numbers if isPrime(i)]
print(prime_random_numbers)
prime_count = len(prime_random_numbers)
print(prime_count)

most_freq_number = max(set(random_numbers), key = random_numbers.count)
max_count = random_numbers.count(most_freq_number)
print(f"most frquent number is : {most_freq_number}")
print(f"and max count is : {max_count}")
```

Key Difference: Passing a Function vs. Calling a Function
random_numbers.count(x) (With parentheses and argument):

Calls the function right now with x and returns an integer count (e.g., 3).

random_numbers.count (Without parentheses):

Refers to the function itself as a tool (a reference), which can be handed over to another function like max().

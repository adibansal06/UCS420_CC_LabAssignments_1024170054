## Q1. Sensor Readings

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
print("Corrected temperature readings:", corrected)

fahrenheit = (9 / 5) * corrected + 32
print("Temperature in Fahrenheit:", fahrenheit)

greater_than_32 = corrected[corrected > 32]
print("Readings greater than 32°C:", greater_than_32)

count = np.sum(corrected > 32)
print("Number of readings greater than 32°C:", count)
```

Vectorization performs an operation on the complete array at once instead of processing each value separately. Boolean indexing directly selects the values satisfying a condition. Both approaches reduce the need for explicit loops and make NumPy operations faster and cleaner for large datasets.

## Q2. Daily Steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", max_position)
print("Maximum value:", steps[max_position])
```

## Q3. NumPy Slicing, Copying and Reshaping

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original after modifying copied array:", original)
print("Copied array:", copied_array)
```

```python
arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print("3 × 4 matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1–2 and columns 2–3:")
print(matrix[0:2, 1:3])
```

```python
flattened = matrix.flatten()
raveled = matrix.ravel()

print("Using flatten():", flattened)
print("Using ravel():", raveled)

raveled[0] = 100
print("Matrix after modifying ravel():")
print(matrix)

flattened[1] = 200
print("Matrix after modifying flatten():")
print(matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q4. Cognitive-Assistive System and OLS

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
XT = X.T
print("X.T:")
print(XT)

print("X.T changes the rows into columns and the columns into rows. Each row of XT represents one feature across all users.")
```

```python
XTX = X.T @ X
print("X.T @ X:")
print(XTX)
```

```python
XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inverse)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("OLS coefficients:", beta)
```

The three coefficients correspond to sleep hours, activity level, and stress level respectively. Each coefficient indicates how the predicted assistance score changes with a one-unit change in that feature while the other features are considered in the model.

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```

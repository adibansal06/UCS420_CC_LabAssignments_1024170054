# Q2 (b)

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

print("\nQ2(b)(i)")
x_values, x_counts = np.unique(x, return_counts=True)
most_frequent_x = x_values[np.argmax(x_counts)]
print("Most frequent value:", most_frequent_x)
print("Indices:", np.where(x == most_frequent_x)[0])

print("\nQ2(b)(ii)")
y_values, y_counts = np.unique(y, return_counts=True)
most_frequent_y = y_values[np.argmax(y_counts)]
print("Most frequent value:", most_frequent_y)
print("Indices:", np.where(y == most_frequent_y)[0])
```

# Q4

```python
Aditya_Bansal = np.linspace(10, 100, 25)

print("\nQ4")
print("Array:", Aditya_Bansal)
print("Dimensions:", Aditya_Bansal.ndim)
print("Shape:", Aditya_Bansal.shape)
print("Total elements:", Aditya_Bansal.size)
print("Data type:", Aditya_Bansal.dtype)
print("Total bytes:", Aditya_Bansal.nbytes)

transpose_Aditya_Bansal = Aditya_Bansal.reshape(25, 1)
print("Transpose using reshape():")
print(transpose_Aditya_Bansal)

print("Transpose using T attribute:")
print(Aditya_Bansal.T)

# Yes, T can also be used, but for a 1-D array it does not change its shape.
# reshape(25, 1) converts it into a column array.
```

# Q5

```python
ucs420_Aditya_Bansal = np.array([[10, 20, 30, 40],
                                 [50, 60, 70, 80],
                                 [90, 15, 20, 35]])

print("\nQ5")
print("Original array:")
print(ucs420_Aditya_Bansal)

print("Mean:", np.mean(ucs420_Aditya_Bansal))
print("Median:", np.median(ucs420_Aditya_Bansal))
print("Maximum:", np.max(ucs420_Aditya_Bansal))
print("Minimum:", np.min(ucs420_Aditya_Bansal))
print("Unique elements:", np.unique(ucs420_Aditya_Bansal))

reshaped_ucs420_Aditya_Bansal = ucs420_Aditya_Bansal.reshape(4, 3)
print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Aditya_Bansal)

resized_ucs420_Aditya_Bansal = np.resize(ucs420_Aditya_Bansal, (2, 3))
print("\nResized array (2 x 3):")
print(resized_ucs420_Aditya_Bansal)
```

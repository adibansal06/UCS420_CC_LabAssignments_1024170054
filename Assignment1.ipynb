```python
def oddSum(n):
  if n%2 == 0:
    return (n//2)*(n//2)
  else:
    return ((n+1)//2)*((n+1)//2)

print(oddSum(9))
```

```python
def primeSum(n):
  s = 0
  for j in range(2, n+1):
    f = 0
    for i in range(2, j//2 +1):
      if j%i == 0:
        f = 1
        break
    if f == 0:
      s = s+j
  return s

print(primeSum(100))
```
